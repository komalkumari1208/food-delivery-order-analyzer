```python
total_orders = df['order_id'].nunique()
total_revenue = df['cost_of_the_order'].sum()
average_order_value = df['cost_of_the_order'].mean()
average_rating = df['rating'].mean()
average_delivery_time = df['delivery_time'].mean()

print("Total Orders:", total_orders)
print("Total Revenue:", round(total_revenue, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Average Rating:", round(average_rating, 2))
print("Average Delivery Time:", round(average_delivery_time, 2), "minutes")
```

```python
restaurant_orders = (
    df.groupby('restaurant_name')['order_id']
      .count()
      .sort_values(ascending=False)
)

restaurant_orders
```

```python
top_restaurants = restaurant_orders.head(10)

top_restaurants
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))

top_restaurants.plot(kind='bar')

plt.title('Top 10 Restaurants by Number of Orders')
plt.xlabel('Restaurant')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
restaurant_revenue = (
    df.groupby('restaurant_name')['cost_of_the_order']
      .sum()
      .sort_values(ascending=False)
)

restaurant_revenue.head(10)
```

```python
plt.figure(figsize=(10,5))

restaurant_revenue.head(10).plot(kind='bar')

plt.title('Top 10 Restaurants by Revenue')
plt.xlabel('Restaurant')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
cuisine_orders = (
    df.groupby('cuisine_type')['order_id']
      .count()
      .sort_values(ascending=False)
)

cuisine_orders
```

```python
plt.figure(figsize=(10,5))

cuisine_orders.plot(kind='bar')

plt.title('Orders by Cuisine')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
location_orders = (
    df.groupby('location')['order_id']
      .count()
      .sort_values(ascending=False)
)

location_orders
```

```python
plt.figure(figsize=(8,5))

location_orders.plot(kind='bar')

plt.title('Orders by Location')
plt.xlabel('Location')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
avg_delivery_by_cuisine = (
    df.groupby('cuisine_type')['delivery_time']
      .mean()
      .sort_values(ascending=False)
)

avg_delivery_by_cuisine
```

```python
plt.figure(figsize=(10,5))

avg_delivery_by_cuisine.plot(kind='bar')

plt.title('Average Delivery Time by Cuisine')
plt.xlabel('Cuisine Type')
plt.ylabel('Average Delivery Time (minutes)')
plt.xticks(rotation=45)
plt.show()
```

```python
day_analysis = (
    df.groupby('day_type')
      .agg(
          Orders=('order_id', 'count'),
          Revenue=('cost_of_the_order', 'sum'),
          Average_Order_Value=('cost_of_the_order', 'mean')
      )
)

day_analysis
```

```python
plt.figure(figsize=(7,5))

day_analysis['Orders'].plot(kind='bar')

plt.title('Weekday vs Weekend Orders')
plt.xlabel('Day Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
df['month'] = df['order_date'].dt.to_period('M').astype(str)
```

```python
monthly_orders = (
    df.groupby('month')['order_id']
      .count()
)

monthly_orders
```

```python
plt.figure(figsize=(10,5))

monthly_orders.plot(kind='line', marker='o')

plt.title('Monthly Order Trend')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
monthly_revenue = (
    df.groupby('month')['cost_of_the_order']
      .sum()
)

monthly_revenue
```

```python
plt.figure(figsize=(10,5))

monthly_revenue.plot(kind='line', marker='o')

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
print("Total Orders:", df['order_id'].nunique())
print("Total Revenue:", round(df['cost_of_the_order'].sum(), 2))
print("Average Order Value:", round(df['cost_of_the_order'].mean(), 2))
print("Average Rating:", round(df['rating'].mean(), 2))
print("Average Delivery Time:", round(df['delivery_time'].mean(), 2))
print("Most Popular Cuisine:", cuisine_orders.idxmax())
print("Top Restaurant by Revenue:", restaurant_revenue.idxmax())
print("Highest Delivery Time Cuisine:", avg_delivery_by_cuisine.idxmax())
print("Highest Order Location:", location_orders.idxmax())
```
